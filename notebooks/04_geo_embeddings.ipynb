# Geo Embedding Similarity Showcase

This notebook demonstrates how to use the `ModalitySimilarity` class to compute similarity between geospatial coordinates.

We use a sinusoidal encoding to map (latitude, longitude) to a 3D vector space, approximating relative distances.

```python
%load_ext autoreload
%autoreload 2
import sys
sys.path.append('../src')

from embedding_similarity.similarity_finder import ModalitySimilarity
import numpy as np
```

```python
sim = ModalitySimilarity()
```

```python
# Define locations (Lat, Lon)
locations = {
    "New York": (40.7128, -74.0060),
    "London": (51.5074, -0.1278),
    "Paris": (48.8566, 2.3522),
    "Tokyo": (35.6762, 139.6503),
    "New Jersey": (40.0583, -74.4057) # Close to NY
}
```

```python
# Compare New York and New Jersey
emb_ny = sim.extract_geo_embedding(*locations["New York"])
emb_nj = sim.extract_geo_embedding(*locations["New Jersey"])

score = sim.compute_similarity(emb_ny, emb_nj)
print(f"Similarity NY <-> NJ: {score:.4f}")
```

```python
# Compare New York and Tokyo
emb_tokyo = sim.extract_geo_embedding(*locations["Tokyo"])
score = sim.compute_similarity(emb_ny, emb_tokyo)
print(f"Similarity NY <-> Tokyo: {score:.4f}")
```
